```python
df_bank2["date"] = pd.to_datetime(df_bank2["date"], errors="coerce")
print(df_bank2["age"].describe())
df_bank2.loc[(df_bank2["age"] < 18) | (df_bank2["age"] > 100), "age"] = np.nan
```

```python
df_bank2["contactado_antes"] = (df_bank2["pdays"] != 999).astype(int)
df_bank2["pdays"] = df_bank2["pdays"].replace(999, np.nan)
```

```python
for c in ["age", "euribor3m", "cons.price.idx"]:
    df_bank2[f"{c}_era_nulo"] = df_bank2[c].isna()
```

```python
df_bank2["age"] = df_bank2["age"].fillna(df_bank2.groupby("job")["age"].transform("median"))
df_bank2["age"] = df_bank2["age"].fillna(df_bank2["age"].median())

# euribor3m y cons.price.idx: mediana por periodo (respaldo: mediana global)
for c in ["euribor3m", "cons.price.idx"]:
    df_bank2[c] = df_bank2[c].fillna(
        df_bank2.groupby(["nr.employed", "emp.var.rate"])[c].transform("median"))
    df_bank2[c] = df_bank2[c].fillna(df_bank2[c].median())

# education: moda por job (respaldo: moda global)
df_bank2["education"] = df_bank2["education"].fillna(
    df_bank2.groupby("job")["education"].transform(
        lambda s: s.mode().iloc[0] if not s.mode().empty else np.nan))
df_bank2["education"] = df_bank2["education"].fillna(df_bank2["education"].mode()[0])

# categóricas: "unknown" como categoría
for c in ["job"]:
    df_bank2[c] = df_bank2[c].fillna("unknown")
    

# marital: pocas filas, se eliminan
df_bank2 = df_bank2.dropna(subset=["marital"]).reset_index(drop=True)
print(df_bank2["marital"].isna().sum(), len(df_bank2))
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
hojas = pd.read_excel('C:\\Users\\carol\\Desktop\\CHARO\\Analisis de Datos\\Python\\Python_for_data\\Proyecto\\customer-details.xlsx', sheet_name=None)  
print(list(hojas.keys()))
print({k: v.shape for k, v in hojas.items()})

cols2 = [set(v.columns) for v in hojas.values()]
print("Columnas iguales:", all(c == cols2[0] for c in cols2))

clientes = pd.concat(
    [d.assign(hoja=nombre) for nombre, d in hojas.items()],
    ignore_index=True)
print(clientes.shape)
```
